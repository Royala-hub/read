# Task 1

```python
n=int(input("Please enter your grade: "))
```

```python
if 91<=n<=100:
    print("A")
elif 81<=n<=90:
    print("B")
elif 71<=n<=80:
    print("C")
elif 61<=n<=70:
    print("D")
elif 51<=n<=60:
    print("E")
elif 0<=n<=50:
    print("F")
else:
    print("Please enter the valid number!")
```

# Task 2

```python
n=int(input("Please enter the positive integer: "))
```

```python
for i in range(1,n+1):
    if i%3==0 and i%5==0:
        print("3, 5")
    elif i%3==0:
        print("3")
    elif i%5==0:
        print("5")
    else:
        print(i)
```
